```python
# 任务1 score_level函数
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数在0-100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

# 输出
print(score_level(95))
print(score_level(85))
print(score_level(70))
print(score_level(50))
# print(score_level(105)) # 取消注释会抛出异常
```

```python
# 任务3 递归版、迭代版阶乘
def fact_iter(n):
    res = 1
    for i in range(2, n+1):
        res *= i
    return res

def fact_recur(n):
    if n == 0:
        return 1
    return n * fact_recur(n-1)

for n in [0,1,5,10]:
    print(f"n={n},迭代阶乘={fact_iter(n)},递归阶乘={fact_recur(n)}")
```

```python
# 任务4 summary函数，返回最小、最大、平均值
def summary(values):
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values)/len(values)
    return (min_v, max_v, avg_v)

data = [10,20,30,40]
min_data, max_data, avg_data = summary(data)
print("最小：",min_data,"最大：",max_data,"平均值：",avg_data)
```

```python
import os
import pickle


print("当前工作目录：", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("数据目录路径：", data_dir)
```

```python
txt_path = os.path.join(data_dir, "students.txt")

# 写入文本内容
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(content)
```

```python
# 任务3：解析函数 parse_student 
def parse_student(line):
    line = line.strip()
    parts = line.split(",")
    # 检查字段数量
    if len(parts) != 2:
        raise ValueError("字段数量不对，需要name,score两个字段")
    name, score_str = parts
    # 姓名不能为空
    if not name.strip():
        raise ValueError("姓名不能为空")
    # 转为整数
    try:
        score = int(score_str.strip())
    except ValueError:
        raise ValueError("成绩不是合法整数")
    # 分数范围校验
    if not (0 <= score <= 100):
        raise ValueError("成绩必须在0~100之间")
    return {"name": name, "score": score}
```

```python
# 任务4：逐行读取，捕获异常，收集有效记录
valid_records = []
error_list = []

with open(txt_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

for line_idx, raw_line in enumerate(lines, start=1):
    raw_line = raw_line.strip()
    if line_idx == 1:
        continue  # 跳过表头
    if not raw_line:
        continue
    try:
        stu_dict = parse_student(raw_line)
        valid_records.append(stu_dict)
    except ValueError as err:
        error_list.append( (line_idx, str(err)) )

print("有效记录：", valid_records)
print("无效记录(行号+原因)：", error_list)

# 统计指标
valid_cnt = len(valid_records)
invalid_cnt = len(error_list)
if valid_cnt > 0:
    avg_score = sum(s["score"] for s in valid_records) / valid_cnt
    pass_count = sum(1 for s in valid_records if s["score"] >=60)
    pass_rate = pass_count / valid_cnt
else:
    avg_score = 0
    pass_rate = 0
```

```python
# 任务5：写入report.txt，演示w和a模式
report_path = os.path.join(data_dir, "report.txt")
report_text = f"""有效记录数：{valid_cnt}
无效记录数：{invalid_cnt}
平均分：{avg_score:.2f}
及格率：{pass_rate:.2%}
---------------------------------------
"""
# w模式：覆盖原有文件
with open(report_path, "w", encoding="utf-8") as f:
    f.write(report_text)
print("w模式写入完成，覆盖旧内容")

# a模式：追加内容，不会删除原有
with open(report_path, "a", encoding="utf-8") as f:
    f.write("【追加一行】a模式：在文件末尾新增文字，不覆盖原有数据\n")
print("a模式追加完成")
```

```python
# 任务6 pickle序列化保存与读取
pkl_path = os.path.join(data_dir, "students.pkl")
# 保存
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)
# 读取
with open(pkl_path, "rb") as f:
    loaded_data = pickle.load(f)

print("pickle读取的数据：", loaded_data)
print("加载数据与原有效记录是否相等：", loaded_data == valid_records)
```

实验 5：类、继承与对象复制

```python
import copy

# 任务1：定义Student类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    
    def is_passed(self):
        return self.score >= 60
    
    def __str__(self):
        return f"学生(name={self.name}, score={self.score})"
```

```python
# 任务2：从实验4有效记录创建Student对象
print("\n实验5 任务2")
stu_obj_list = []
for record in valid_records:
    s = Student(record["name"], record["score"])
    stu_obj_list.append(s)
    print(s, " 及格状态：", s.is_passed())
```

```python
# 任务3：子类 GraduateStudent 继承Student
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic
    
    def __str__(self):
        return f"研究生(name={self.name}, score={self.score}, 研究方向={self.research_topic})"

# 测试子类
grad_stu = GraduateStudent("Tom", 88, "机器学习")
print("\n任务3 继承测试")
print(grad_stu, "及格状态：", grad_stu.is_passed())
```

```python
# 任务4 浅拷贝copy.copy 深拷贝copy.deepcopy对比
print("\n任务4 对象拷贝对比")
# 原始嵌套对象
original = {"name": "Alice", "scores": [85, 90]}

# ①直接赋值，只是同一个对象别名
alias = original
alias["scores"][0] = 99
print("直接赋值后 original:", original)
# 重置
original = {"name": "Alice", "scores": [85, 90]}

# ②浅拷贝 copy.copy
shallow = copy.copy(original)
shallow["scores"][0] = 99
print("浅拷贝后 original:", original)
#重置
original = {"name": "Alice", "scores": [85, 90]}

#③深拷贝 deepcopy
deep = copy.deepcopy(original)
deep["scores"][0] = 99
print("深拷贝后 original:", original)
```
