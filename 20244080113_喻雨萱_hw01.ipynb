```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 测试三组数字
print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
```

如果传入负数，取模运算后，最终返回的仍然是正数最大公约数，GCD 一般只针对非负整数，负数的 GCD 本身没有实际意义。

```python
# 1. euler_limit  欧拉数极限
def euler_limit(n):
    return (1 + 1 / n) ** n

print(euler_limit(4))
```

```python
import math
# 2. euler_infinite_sum 前n项和
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    for k in range(n):
        total += 1 / math.factorial(k)
    return total

print(euler_infinite_sum(4))
print(euler_infinite_sum(0))
print(euler_infinite_sum(1))

```

```python
import math
# 3. euler_approx
def euler_approx(eps):
    true_e = math.exp(1)
    k = 0
    current_sum = 0
    while abs(current_sum - true_e) >= eps:
        current_sum += 1 / math.factorial(k)
        k += 1
    return current_sum

print(euler_approx(0.001))
```

```python
# 4. print_euler_sum_table，打印k从1~n的euler_infinite_sum(k)
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, euler_infinite_sum(k)={val:.6f}")

print_euler_sum_table(5)
```
