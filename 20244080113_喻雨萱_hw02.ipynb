```python
# 1. 判断回文 is_palindrome
def is_palindrome(s):
    # 去掉空格，所有字符全部转小写
    clean = s.replace(" ", "").lower()
    return clean == clean[::-1]
print (is_palindrome("T A C O cat"))
print (is_palindrome("dog"))


```

```python
# 2. 判断字母顺序 is_abecedarian
def is_abecedarian(s):
    # 去掉空格，全部转小写
    clean = s.replace(" ", "").lower()
    for i in range(len(clean)-1):
        if clean[i] > clean[i+1]:
            return False
    return True
print(is_abecedarian("abcd efgh xyz")) 
print(is_abecedarian("dog eun")) 
```

```python
# 3. 删除元音 remove_vowels
def remove_vowels(s):
    vowels = {'a','e','i','o','u','A','E','I','O','U'}
    res = ""
    for c in s:
        if c not in vowels:
            res += c
    # 如果全是元音，显示空
    if res == "":
        res = "空"
    return res

print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("AEO"))
```

```python
# 1. 列表反转 list_reverse
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]


# 测试
print(list_reverse([1,2,3]))

try:
    print(list_reverse(123))
except TypeError as err:
    print("捕获错误：", err)
```

```python
# 2. is_sorted，seq只遍历一次
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True
print(is_sorted([1,2,4,5]))
print(is_sorted([2,1,3]))
```

```python
# 1. 向量数乘 vec_scalar_mult
def vec_scalar_mult(t, s):
    # 检查t是不是元组
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    # 检查s是不是int或float
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或浮点数")
    # 检查元组里面每个元素都是数字
    for item in t:
        if not isinstance(item, (int, float)):
            raise TypeError("元组内元素只能是整数或浮点数")
    # 每个元素乘s，返回新元组
    return tuple(x * s for x in t)
print(vec_scalar_mult((1, 2, 3), 2))
try:
    print(vec_scalar_mult((1, 2, 3), (2,3)))
except TypeError as e:
    print("报错：", e)
```

```python
# 2. 向量内积 vec_inner_product
def vec_inner_product(v1, v2):
    if not isinstance(v1, tuple) or not isinstance(v2, tuple):
        raise TypeError("输入必须是元组")
    if len(v1) != len(v2):
        raise ValueError("两个向量长度不一致")
    total = 0.0
    for a, b in zip(v1, v2):
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise TypeError("向量元素只能是整数或浮点数")
        total += a * b
    return total
print(vec_inner_product((1,2,3), (4,5,6)))
try:
    print(vec_inner_product((1,2), (3,4,5)))
except ValueError as e:
    print("报错：", e)

try:
    print(vec_inner_product([1,2], (3,4)))
except TypeError as e:
    print("报错：", e)
```

```python
# 3. 检查有效矩阵 check_valid_mx
def check_valid_mx(my_mx):
    # 外层必须是元组
    if not isinstance(my_mx, tuple):
        return False
    # 空矩阵也算有效
    row_len = None
    for row in my_mx:
        # 每一行必须是元组
        if not isinstance(row, tuple):
            return False
        # 记录第一行长度，后面每行和它对比
        if row_len is None:
            row_len = len(row)
        else:
            if len(row) != row_len:
                return False
        # 检查每个元素是int/float
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True
mx_test = ((1,2), (3,4))
print(check_valid_mx(mx_test))
mx_bad1 = ((1,2), (3,"4"))
print(check_valid_mx(mx_bad1))
```

```python
# 4. 矩阵乘向量 mx_vec_mult
def mx_vec_mult(mx, vec):
    #用check_valid_mx校验矩阵
    if not check_valid_mx(mx):
        raise TypeError("输入不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    # 取出矩阵的列数 = 向量长度，不匹配报错
    if len(mx) > 0 and len(mx[0]) != len(vec):
        raise ValueError("矩阵列数和向量长度不一致")
    # 矩阵 × 向量：每行和向量做内积
    res = []
    for row in mx:
        res.append(vec_inner_product(row, vec))
    return tuple(res)
print(mx_vec_mult(mx_test, (1, 2)))
try:
    print(mx_vec_mult(mx_test, [1,2]))
except TypeError as e:
    print("报错：", e)
```

```python
class Vector:
    def __init__(self, dim, elements=None):
        # 判断维度是否为整数
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        # 判断维度非负
        if dim < 0:
            raise ValueError("维度不能是负数")

        # 不传入元素，默认全0向量
        if elements is None:
            self.elements = tuple([0.0 for _ in range(dim)])
        else:
            elements = tuple(elements)
            # 元素数量和维度匹配校验
            if len(elements) != dim:
                raise ValueError("元素个数和维度不一致")
            # 校验每个元素是int或者float
            for item in elements:
                if not isinstance(item, (int, float)):
                    raise TypeError("向量元素只能是整数或浮点数")
            self.elements = elements
        self.dim = dim


v0 = Vector(3)
print(v0.dim, v0.elements)

v1 = Vector(2, (2,4))
print(v1.dim, v1.elements)
```

#2
维度 dim 和元素 elements 应该是实例属性。因为每一个 Vector 实例代表一个独立向量，不同向量可以拥有不同维度、不同元素；类属性是所有实例共享的数据，在这里不合适。。

```python
# 3.比较运算符
    # 给已经定义好的Vector类追加方法，顶格写没问题
def __eq__(self, other):
    if not isinstance(other, Vector):
        return False
    if self.dim != other.dim:
        raise ValueError("不能比较维度不同的向量")
    return self.elements == other.elements

def __lt__(self, other):
    if not isinstance(other, Vector):
        raise TypeError("只能和Vector对象比较")
    if self.dim != other.dim:
        raise ValueError("不能比较维度不同的向量")
    return self.elements < other.elements

def __le__(self, other):
    return self < other or self == other

def __gt__(self, other):
    return not self <= other

def __ge__(self, other):
    return not self < other
# 把函数绑定Vector
Vector.__eq__ = __eq__
Vector.__lt__ = __lt__
Vector.__le__ = __le__
Vector.__gt__ = __gt__
Vector.__ge__ = __ge__

v1 = Vector(2, (2,4))
v2 = Vector(2, (2,5))
print(v1 < v2)
print(v1 == v2)
```

```python
# 4. dot内积
def dot(self, other):
    if not isinstance(other, Vector):
        raise TypeError("dot的参数必须是Vector对象")
    if self.dim != other.dim:
        raise ValueError("两个向量维度不一致，无法计算内积")
    return sum(a * b for a, b in zip(self.elements, other.elements))
# 5. 乘法
def __mul__(self, other):
    # 向量 * 向量
    if isinstance(other, Vector):
        if self.dim != other.dim:
            raise ValueError("向量维度不一致，不能逐元素相乘")
        new_ele = tuple(a * b for a, b in zip(self.elements, other.elements))
        return Vector(self.dim, new_ele)
    # 向量 * 标量
    elif isinstance(other, (int, float)):
        new_ele = tuple(x * other for x in self.elements)
        return Vector(self.dim, new_ele)
    else:
        raise TypeError("不支持该类型乘法")

def __rmul__(self, other):
    return self * other
def __repr__(self):
    return f"Vector({self.dim}, {self.elements})"

# 绑定到Vector类
Vector.dot = dot
Vector.__mul__ = __mul__
Vector.__rmul__ = __rmul__
Vector.__repr__ = __repr__

v3 = Vector(2, (2, 3))
v4 = Vector(2, (4, 5))
print("内积：", v3.dot(v4))
print("向量相乘 v3*v4：", v3 * v4)
print("v3 * 2：", v3 * 2)
print("2 * v3：", 2 * v3)
```
